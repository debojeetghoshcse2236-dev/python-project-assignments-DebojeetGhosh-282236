1. Given two sets, A = {1, 2, 3, 4, 5} and B = {4, 5, 6, 7, 8}, display their union, intersection, and difference (A - B).

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference (A - B):", A - B)
```

2. Common Elements in Two Lists
Write a program to accept two lists of integers and display all distinct elements that appear in both lists using sets.

```python
list1 = []
list2 = []

n = int(input("Enter number of elements in first list: "))
for i in range(n):
    x = int(input("Enter element: "))
    list1.append(x)

m = int(input("Enter number of elements in second list: "))
for i in range(m):
    x = int(input("Enter element: "))
    list2.append(x)

set1 = set(list1)
set2 = set(list2)

common = set1 & set2

print("Common elements:", common)
```

3. Check Subset and Superset
Given two sets, write a program to check whether the first set is a subset of the second set and whether the second set is a superset of the first.

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

if A.issubset(B):
    print("A is a subset of B")
else:
    print("A is not a subset of B")

if B.issuperset(A):
    print("B is a superset of A")
else:
    print("B is not a superset of A")
```

4. Find Students Enrolled in Exactly One Course
Two sets represent students enrolled in Python and Java courses.
python_students = {"Aman", "Riya", "Soham", "Priya"}
java_students = {"Riya", "Priya", "Rahul", "Ankit"}
Write a program to display students enrolled in exactly one of the two courses, excluding those enrolled in both.


```python
python_students = {"Aman", "Riya", "Soham", "Priya"}
java_students = {"Riya", "Priya", "Rahul", "Ankit"}

result = python_students ^ java_students

print("Students enrolled in exactly one course:", result)
```
